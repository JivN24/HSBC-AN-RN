```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

DL = '/Users/aksharaa/Downloads/'

# multi-tenor implied vol
vol = pd.read_excel(DL + 'diff_vols.xlsx')
vol.columns = ['date', 'v1w', 'v2w', 'v1m']          # rename positionally (kills the stray space)
vol['date'] = pd.to_datetime(vol['date'])
vol = vol.dropna().sort_values('date').reset_index(drop=True)

# merged frame with the event flag
ev = pd.read_csv(DL + 'usdjpy_merged_1.csv', parse_dates=['date'])
ev = ev.rename(columns={'fomc': 'event'})            # use 'event' generically; add BoJ later if you want
print('vol:', vol.shape, '| events frame:', ev.shape, '| event days:', ev['event'].sum())
```

```python
df = vol.merge(ev[['date', 'spot', 'event']], on='date', how='inner').reset_index(drop=True)
df.to_csv(DL + 'tenors_with_events.csv', index=False)   # saved merged frame
print(df.shape, '| event days in merged frame:', df['event'].sum())
df.head()
```

```python
TENORS = np.array([7, 14, 30])   # calendar days for 1W, 2W, 1M

def event_maturity_vol(row, dte):
    """Vol of an option maturing 'dte' calendar days from now."""
    ys = np.array([row['v1w'], row['v2w'], row['v1m']])
    if dte <= 7:  return row['v1w']    # can't go shorter than 1W -> flat
    if dte >= 30: return row['v1m']    # cap at 1M
    return np.interp(dte, TENORS, ys)
```

```python
def rolldown(target_dates, max_dte=30):
    rows = []
    for E in pd.to_datetime(target_dates):
        for _, r in vol.iterrows():
            dte = (E - r['date']).days
            if 0 < dte <= max_dte:
                rows.append((dte, event_maturity_vol(r, dte)))
    d = pd.DataFrame(rows, columns=['dte', 'v'])
    g = d.groupby('dte')['v'].agg(['mean', 'std', 'count']).reset_index()
    g['se'] = g['std'] / np.sqrt(g['count'])
    return g

events = df.loc[df['event'] == 1, 'date'].tolist()
curve_event = rolldown(events)
print(f'{len(events)} events')
curve_event[['dte', 'mean', 'se', 'count']].head()
```

```python
rng = np.random.default_rng(0)
def far_from_events(t): return all(abs((t - e).days) >= 45 for e in pd.to_datetime(events))
candidates = [t for t in vol['date'] if far_from_events(t)]
pseudo = list(rng.choice(candidates, size=len(events), replace=True))
curve_pseudo = rolldown(pseudo)
```

```python
plt.figure(figsize=(10,6))
for g, lab, col in [(curve_event,  f'Toward FOMC events (n={len(events)})', '#c0392b'),
                    (curve_pseudo, f'Toward non-event days (n={len(pseudo)})', '#2c3e50')]:
    plt.plot(g['dte'], g['mean'], marker='o', ms=4, label=lab, color=col)
    plt.fill_between(g['dte'], g['mean']-g['se'], g['mean']+g['se'], alpha=0.15, color=col)
plt.gca().invert_xaxis()   # count DOWN toward the event (left-to-right approaches day 0)
plt.xlabel('Calendar days until the date  (approaching →)')
plt.ylabel('Implied vol of option maturing on that date')
plt.title('USDJPY: event-maturity implied vol approaching the date')
plt.legend(); plt.grid(alpha=0.3); plt.tight_layout(); plt.show()
```
