```python
import pandas as pd, numpy as np
import matplotlib.pyplot as plt
from statsmodels.tsa.stattools import grangercausalitytests
from sklearn.feature_selection import mutual_info_regression

df = pd.read_csv('/Users/aksharaa/Downloads/crosspair_daily.csv', parse_dates=['date'])
pairs = ['ret_eurusd', 'ret_usdsgd', 'ret_usdmxn', 'ret_usdpln']
lbl = {p: p.replace('ret_','').upper() for p in pairs}
print(df.shape); df[pairs].head()
```

```python
maxlag = 5
def gp(x, y):
    # does x lead y? statsmodels tests col2 -> col1, so pass [y, x]
    res = grangercausalitytests(df[[y, x]].dropna(), maxlag=maxlag, verbose=False)
    return min(res[k][0]['ssr_ftest'][1] for k in res)   # best p across lags 1..maxlag

print(f"{'leader':8s} -> {'follower':8s}   Granger p")
print("-"*36)
for x in pairs:
    for y in pairs:
        if x != y:
            p = gp(x, y)
            star = ' *' if p < 0.05 else ''
            print(f"{lbl[x]:8s} -> {lbl[y]:8s}   {p:.4f}{star}")
```

```python
print("Net lead-lag verdict per pair:\n")
for i, x in enumerate(pairs):
    for y in pairs[i+1:]:
        pxy, pyx = gp(x, y), gp(y, x)
        if   pxy < 0.05 and pyx >= 0.05: verdict = f"{lbl[x]} leads {lbl[y]}"
        elif pyx < 0.05 and pxy >= 0.05: verdict = f"{lbl[y]} leads {lbl[x]}"
        elif pxy < 0.05 and pyx < 0.05:  verdict = f"bidirectional ({lbl[x]} stronger)" if pxy<pyx else f"bidirectional ({lbl[y]} stronger)"
        else:                            verdict = "no lead either way"
        print(f"  {lbl[x]:7s} vs {lbl[y]:7s}: {verdict}")
```

```python
def tdmi(x, y, lags, n_shuffle=150, seed=0):
    rng = np.random.default_rng(seed)
    x, y = np.asarray(x), np.asarray(y)
    mi, null_hi = [], []
    for k in lags:
        xl, yk = x[:-k], y[k:]
        mi.append(mutual_info_regression(xl.reshape(-1,1), yk, random_state=seed)[0])
        nulls = [mutual_info_regression(rng.permutation(xl).reshape(-1,1), yk, random_state=seed)[0]
                 for _ in range(n_shuffle)]
        null_hi.append(np.percentile(nulls, 95))
    return np.array(mi), np.array(null_hi)

lags = range(1, 11)
# USD-driver framing: test EURUSD (USD proxy) leading each of the others
drivers = [('ret_eurusd','ret_usdsgd'), ('ret_eurusd','ret_usdmxn'), ('ret_eurusd','ret_usdpln')]

fig, ax = plt.subplots(1, 3, figsize=(15,4), sharey=True)
for a,(x,y) in zip(ax, drivers):
    mi, nh = tdmi(df[x].values, df[y].values, lags)
    a.plot(list(lags), mi, marker='o', label='TDMI')
    a.plot(list(lags), nh, '--', color='grey', label='95% null')
    a.set_title(f'{lbl[x]} → {lbl[y]}'); a.set_xlabel('lag (days)'); a.legend()
ax[0].set_ylabel('mutual information')
plt.tight_layout(); plt.show()
```

```python
def xcorr(x, y, mx=10):
    ks = range(-mx, mx+1)
    return list(ks), [np.corrcoef(x[:len(x)-k], y[k:])[0,1] if k>=0
                      else np.corrcoef(x[-k:], y[:len(y)+k])[0,1] for k in ks]

fig, ax = plt.subplots(1, 3, figsize=(15,4))
for a,(x,y) in zip(ax, drivers):
    ks, cs = xcorr(df[x].values, df[y].values)
    a.bar(ks, cs); a.axvline(0, color='k', lw=0.8)
    a.set_title(f'{lbl[x]} (t) vs {lbl[y]} (t+k)'); a.set_xlabel('lag k')
plt.tight_layout(); plt.show()
```
